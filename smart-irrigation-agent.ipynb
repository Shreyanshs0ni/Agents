# Question


Design a Smart Irrigation Agent for a farm that decides whether to turn the irrigation system ON or OFF based on soil moisture and weather conditions.

The agent should:
1. Perceive soil moisture and weather conditions.
2. Decide an action (turn irrigation ON/OFF).
3. Simulate the interaction between the agent and the environment.
4. Measure performance by saving water when irrigation is not needed.

# Smart Irrigation Agent

```python
import random

class SmartIrrigationAgent:
    def __init__(self, moisture_threshold=40):
        self.moisture_threshold = moisture_threshold
        self.water_saved = 0
        self.total_actions = 0

    def perceive(self):
        soil_moisture = random.randint(0, 100)  # percentage
        weather = random.choice(["sunny", "rainy", "cloudy"])
        return soil_moisture, weather

    def decide(self, soil_moisture, weather):
        if soil_moisture < self.moisture_threshold and weather != "rainy":
            return "ON"
        else:
            return "OFF"

    def act(self, action):
        if action == "OFF":
            self.water_saved += 1
        self.total_actions += 1

    def performance(self):
        return f"Water saved in {self.water_saved} out of {self.total_actions} steps"
```

# Simulation


```python
agent = SmartIrrigationAgent()

for step in range(10):
    soil_moisture, weather = agent.perceive()
    action = agent.decide(soil_moisture, weather)
    agent.act(action)

    print(f"Step {step+1}: Moisture={soil_moisture}%, Weather={weather}, Irrigation={action}")

print("\nPerformance:", agent.performance())
```
