```python
from pyswip import Prolog

prolog = Prolog()

# Add facts
prolog.assertz("parent(john, mary)")
prolog.assertz("parent(mary, alice)")

# Add rule
prolog.assertz("grandparent(X, Y) :- parent(X, Z), parent(Z, Y)")

# Query
result = list(prolog.query("grandparent(john, alice)"))

print(result)    #in Output {} means True (success)
```

```python
#Dynamic Knowledge (Adding Facts at Runtime)
#how to assert and retract facts dynamically.

from pyswip import Prolog

prolog = Prolog()

# Add facts dynamically
prolog.assertz("likes(ram, pizza)")
prolog.assertz("likes(ram, burger)")

# Query
for result in prolog.query("likes(ram, X)"):
    print(result)

# Remove fact
prolog.retract("likes(ram, pizza)")

print("After retract:")
for result in prolog.query("likes(ram, X)"):
    print(result)
```

```python
#Implement graph traversal using Prolog rules inside Python.

from pyswip import Prolog

# Initialize Prolog
prolog = Prolog()

# Define graph edges
prolog.assertz("edge(a, b)")
prolog.assertz("edge(b, c)")
prolog.assertz("edge(c, d)")
prolog.assertz("edge(a, d)")
prolog.assertz("edge(d, e)")

# Define path rules (recursive)
prolog.assertz("path(X, Y) :- edge(X, Y)")
prolog.assertz("path(X, Y) :- edge(X, Z), path(Z, Y)")

# Query all reachable nodes from 'a'
print("Nodes reachable from 'a':")
results = list(prolog.query("path(a, X)"))

# Remove duplicates
unique_results = {res['X'] for res in results}

for node in unique_results:
    print(node)

# Query specific path existence
print("\nCheck if path exists from a to e:")
if list(prolog.query("path(a, e)")):
    print("Path exists!")
else:
    print("No path found.")
```

```python
#Use logical rules to classify students as Pass or Fail based on marks.

from pyswip import Prolog

# Initialize Prolog
prolog = Prolog()

# Add student marks (facts)
prolog.assertz("marks(ram, 75)")
prolog.assertz("marks(shyam, 40)")
prolog.assertz("marks(sita, 60)")
prolog.assertz("marks(gita, 30)")

# Define rules
prolog.assertz("pass(X) :- marks(X, M), M >= 50")
prolog.assertz("fail(X) :- marks(X, M), M < 50")

# Query passed students
print("Passed Students:")
passed = list(prolog.query("pass(X)"))

for student in passed:
    print(student['X'])

# Query failed students
print("\nFailed Students:")
failed = list(prolog.query("fail(X)"))

for student in failed:
    print(student['X'])

# Check specific student result
name = input("\nEnter student name to check result: ")

if list(prolog.query(f"pass({name})")):
    print(f"{name} has PASSED")
elif list(prolog.query(f"fail({name})")):
    print(f"{name} has FAILED")
else:
    print("Student not found")
```

```python
#Build a rule-based chatbot using Prolog logic integrated with Python.

from pyswip import Prolog

# Initialize Prolog
prolog = Prolog()

# Add chatbot knowledge (facts)
prolog.assertz("response(hello, 'Hi! How can I help you?')")
prolog.assertz("response(hi, 'Hello! What do you need?')")
prolog.assertz("response(bye, 'Goodbye! Have a nice day!')")
prolog.assertz("response(thanks, 'You are welcome!')")
prolog.assertz("response(help, 'I can answer basic greetings and simple queries.')")

# Chat loop
print("Simple Chatbot (type 'exit' to stop)")

while True:
    user_input = input("\nYou: ").lower()

    if user_input == "exit":
        print("Bot: Session ended.")
        break

    # Query Prolog
    query = f"response({user_input}, X)"
    result = list(prolog.query(query))

    if result:
        print("Bot:", result[0]['X'])
    else:
        print("Bot: I don't understand. Try saying hello, help, or thanks.")
```

```python
#Perform arithmetic operations using Prolog rules integrated with Python.

from pyswip import Prolog

# Initialize Prolog
prolog = Prolog()

# Define arithmetic rules
prolog.assertz("sum(X, Y, Z) :- Z is X + Y")
prolog.assertz("subtract(X, Y, Z) :- Z is X - Y")
prolog.assertz("multiply(X, Y, Z) :- Z is X * Y")
prolog.assertz("divide(X, Y, Z) :- Y =\\= 0, Z is X / Y")

# Perform operations
print("Arithmetic Operations using PySWIP\n")

# Addition
result = list(prolog.query("sum(10, 5, Z)"))
print("10 + 5 =", result[0]['Z'])

# Subtraction
result = list(prolog.query("subtract(10, 5, Z)"))
print("10 - 5 =", result[0]['Z'])

# Multiplication
result = list(prolog.query("multiply(10, 5, Z)"))
print("10 * 5 =", result[0]['Z'])

# Division
result = list(prolog.query("divide(10, 5, Z)"))
print("10 / 5 =", result[0]['Z'])

# User Input Section
print("\n--- User Input Calculation ---")
x = int(input("Enter first number: "))
y = int(input("Enter second number: "))
op = input("Enter operation (+, -, *, /): ")

if op == "+":
    query = f"sum({x}, {y}, Z)"
elif op == "-":
    query = f"subtract({x}, {y}, Z)"
elif op == "*":
    query = f"multiply({x}, {y}, Z)"
elif op == "/":
    query = f"divide({x}, {y}, Z)"
else:
    print("Invalid operation")
    exit()

result = list(prolog.query(query))

if result:
    print("Result =", result[0]['Z'])
else:
    print("Error (possibly division by zero)")
```
