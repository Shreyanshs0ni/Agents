```python
#EXAMPLE 1: RED, GREEN AND YELLOW TRAFFIC LIGHT

!pip install experta
!pip install frozendict==1.2
import collections
import collections.abc
collections.Mapping = collections.abc.Mapping
collections.MutableMapping = collections.abc.MutableMapping
collections.Sequence = collections.abc.Sequence

from random import choice
from experta import *

class Light(Fact):
  """Info about the traffic light."""
  pass

class RobotCrossStreet(KnowledgeEngine):
  @Rule(Light(color='green'))
  def green_light(self):
      print("Walk")
  @Rule(Light(color='red'))
  def red_light(self):
      print("Don't walk")
  @Rule(AS.light << Light(color=L('yellow') | L('blinking-yellow')))
  def cautious(self, light):
      print("Be cautious because light is", light["color"])
```

```python
engine = RobotCrossStreet()
engine.reset()
engine.declare(Light(color=choice(['green', 'yellow', 'blinking-yellow', 'red'])))
engine.run()
```

```python
#COMPLETE EXAMPLE

from experta import *
#Color is red
#Temperature is high, are written as

Fact(color='red')
Fact(temp='high')

#Creating Fact Classes
class Alert(Fact):
    pass
class Status(Fact):
    pass

Alert('red')
Status('critical')

#Defining the Rules, define what to do when conditions are met and running knowledge engine
class MyEngine(KnowledgeEngine):
  @Rule(Fact(temp='high'))
  def hot(self):
      print("It's hot!")

  @Rule(Fact(temp='low'))
  def low(self):
      print("Temperature is very low!")

  #Defining multiple conditions AND
  @Rule(Fact(temp='high') & Fact(humidity='high'))
  def discomfort(self):
      print("Very uncomfortable weather")

  #Defining multiple conditions OR
 # @Rule(Fact(temp='high') | Fact(temp='low'))
 # def extreme(self):
 #     print("Extreme condition")

engine = MyEngine()
engine.reset()  # clears old facts

#engine.declare(Fact(temp='low'))  # add fact
#engine.declare(Fact(temp='low') | Fact(humidity='high'))  # wrong
engine.declare(Fact(temp='high'))
engine.declare(Fact(humidity='low'))

engine.run()  # execute rules
```
